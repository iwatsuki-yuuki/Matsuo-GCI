```python
from sklearn.model_selection import train_test_split
import lightgbm as lgb
```

```python
X = df.drop('Attrition',axis=1)
y = df['Attrition'] # 目的変数
# トレーニングデータ,テストデータの分割
X_train, X_valid, y_train, y_valid = train_test_split(X, y,test_size=0.2, random_state=0)
```

今回は不均衡データであることから、評価指標にはAUCを用いることにします。適切な評価指標を選ぶことは最終課題でも重要です。

```python
# 学習に使用するデータを設定
lgb_train = lgb.Dataset(X_train, y_train)
lgb_eval = lgb.Dataset(X_valid, y_valid, reference=lgb_train)

# パラメータ
params = {
        'task': 'train',
        'boosting_type': 'gbdt',
        'objective': 'binary', # 目的 : 分類
        'metric': {'auc'},
         'num_leaves': 20,
        'max_depth':5,
        'min_data_in_leaf': 3,
        'num_iteration': 100,
        'learning_rate':0.03,
        'num_boost_round':100,
         'early_stopping_rounds':20,
}

# モデルの学習
model = lgb.train(params,
                  train_set=lgb_train, # トレーニングデータの指定
                  valid_sets=lgb_eval, # 検証データの指定
                  )

# テストデータの予測
y_pred = model.predict(X_valid)
```

ここで成果は出ているが正しく評価されずインセンティブをもらえていない人に注目することにした（もし正しく評価されていなかったら社内評価システムの導入提案などをできるだろうと考えたから）。またこの場合は給与を上げるのではなくインセンティブを上げる人を増やす（0の人が多いので支給額を上げるよりかはもらえるハードルを下げる方が大切だと思った）。

```python
# MonthlyAchievementに以下の閾値を設けてそれより高いかを確認する
q_high  = 0.70     # 上位30%を「高い」
q_low   = 0.30     # 下位30%を「低い」

# 閾値を計算
ma_high = df['MonthlyAchievement'].quantile(q_high)
pi_high = df['PerformanceIndex'].quantile(q_high)
pr_low  = df['PerformanceRating'].quantile(q_low)

# フラグ列を付ける
df['is_high_MA']  = df['MonthlyAchievement'] >= ma_high
df['is_high_PI']  = df['PerformanceIndex']   >= pi_high
df['is_low_PR']   = df['PerformanceRating']  <= pr_low

# 条件に合う行(MonthlyAchievement か PerformanceIndex のどちらかが高い行)を取り出す
mask = (df[['is_high_MA','is_high_PI']].any(axis=1)) & df['is_low_PR']
gap_df = df[mask]

print('該当レコード数:', len(gap_df))
display(gap_df[['EmployeeNumber','MonthlyAchievement',
                'PerformanceIndex','PerformanceRating']].head())

# 集計
pivot = (
    gap_df
    .groupby(['Department','JobLevel'])
    .size()
    .unstack(fill_value=0)
    .sort_index(ascending=False)
)
print(pivot.head())
```

```python
# 後々使うためgap_flag列を作成
df['gap_flag'] = (
        ((df['MonthlyAchievement'] >= ma_high) |   # どちらか高い
         (df['PerformanceIndex']   >= pi_high))
        &
        (df['PerformanceRating']   <= pr_low)      # かつ評価が低い
).astype(int)          # 1=True, 0=False

print('ギャップ対象者数:', df['gap_flag'].sum())
```

## 4.打ち手仮説立案

若手 × 低収入 × 高ストレス × 出張過多の離職率を高めている重要な要因であった。


また若手が評価されていない現状にも問題があると判断した。



- 離職率予測+若者のストレス軽減×出張過多防止
    - リモートツールの導入によってストレスの軽減と出張過多を防ぐ
    - 出張の量を均等にする＋全体の主張回数を減らす
    - リモートによる情報共有不足を解消するシステムも検討？
- 離職率予測＋ストレス軽減＋収入増やす
    - 市場分析の時に離職率を減らすための取り組みとして以下のものが見られた
        1.   給与全体の上昇
        2.   福利厚生や研修制度の充実
    - 実際になされている取り組みである
- 離職率予測＋インセンティブの改善（収入増やす）
    - 正しく評価される人事制度の導入

```python
df_cmp = (df_now[['pred_prob_now']]
          .join(df_after['pred_prob_after']))
df_cmp['delta_prob'] = df_cmp['pred_prob_now'] - df_cmp['pred_prob_after']
```

```python
exp_leave_before = df_cmp['pred_prob_now'].sum()
exp_leave_after  = df_cmp['pred_prob_after'].sum()
prevented_head   = exp_leave_before - exp_leave_after

mean_before = df_cmp['pred_prob_now'].mean()   # 全社員の平均離職確率
mean_after  = df_cmp['pred_prob_after'].mean()
mean_drop   = mean_before - mean_after         # 低減幅（ポイント差）

print(f'防止できる期待離職者数: {prevented_head:.2f} 名')
print(f'全社員比              : {prevented_head/len(df_cmp):.2%}')
print('――――――――――――――――――――――――――――――')
print(f'平均離職確率 施策前   : {mean_before:.2%}')
print(f'平均離職確率 施策後   : {mean_after:.2%}')
print(f'平均離職確率 低減幅   : {mean_drop:.2%}')
```

## 6.打ち手仮説検証2

**離職率予測+若者のストレス軽減×出張過多防止**

若者(Age<=30)のストレスを一段階下げ、BusinessTravelは若者は全体平均にする

```python
# 基準データ
df_base = df.copy()
df_base['pred_prob_now'] = model.predict(df_base[feature_cols])

# 若手用
df_young = df.copy()

# 30歳以下は若者扱い
young_mask = df_young['Age'] <= 30

# StressRating を1下げる
df_young.loc[young_mask, 'StressRating'] = (
    df_young.loc[young_mask, 'StressRating'] - 1
).clip(lower=1)

# BusinessTravelを全社平均に置換
bt_mean = df['BusinessTravel'].mean()
df_young.loc[young_mask, 'BusinessTravel'] = round(bt_mean)

# 推論
df_young['pred_prob_young'] = model.predict(df_young[feature_cols])

import pandas as pd

cmp = pd.DataFrame({
    'base'  : df_base ['pred_prob_now'],
    'young' : df_young['pred_prob_young']
})
cmp['delta_prob'] = cmp['base'] - cmp['young']

# 平均離職確率の変化
mean_drop = cmp['delta_prob'].mean()
print(f'平均離職確率低減幅 : {mean_drop:.2%}')

# 期待離職者数 (確率合計) の減少
prevented = cmp['delta_prob'].sum()
print(f'防止できる期待離職者数  : {prevented:.1f} 名 ({prevented/len(cmp):.2%} 全社員比)')
```

```python
LOSS_PER_LEAVE = 24_000_000        # 2,400 万円 ＝ (年収1,200×1.5)＋採用600
saved_cost = prevented * LOSS_PER_LEAVE    # 円

# ❹ ツール導入コスト（100 万円／月 × 12 か月）
TOOL_COST_ANNUAL = 12_000_000      # 円

# ❺ 純利益 (ROI)
net_profit = saved_cost - TOOL_COST_ANNUAL

# ❻ 結果表示（億円換算）
print('──── ストレス・出張緩和ツール ROI ────')
print(f'期待離職防止人数      : {prevented:.1f} 名 ({prevented/len(cmp):.2%} 全社員比)')
print(f'年間損失削減額        : {saved_cost/1e8:.2f} 億円')
print(f'年間ツール費用        : {TOOL_COST_ANNUAL/1e8:.2f} 億円')
print(f'純利益 (ROI)          : {net_profit/1e8:.2f} 億円')
```
